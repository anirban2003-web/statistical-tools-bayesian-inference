```python
import numpy as np 
import matplotlib.pyplot as plt
#no of points
N= 10000

#to generate random points in first quadrant 
x= np.random.uniform(0,1,N)
y=np.random.uniform(0,1,N)

#to check if the points are inside the circle
inside = x**2+y**2<=1

pi_estimate= 4*np.sum(inside)/N
print(f"Points inside : {np.sum(inside)}")
print(f"Points outside: {N-np.sum(inside)}")
print(f"Estimated pi = {np.round(pi_estimate,5)}")
print(f"True pi= {np.round(np.pi,5)}")
print(f"Error = {pi_estimate-np.pi}")

#plot
fig, ax = plt.subplots(figsize=(6,6))
ax.scatter(x[inside],y[inside],color='green',s=1.5,label = 'Inside')
ax.scatter(x[~inside],y[~inside],color='red',s=1.5,label = 'Outside')
#drawing the quarter circle
theta = np.linspace(0, np.pi/2,500)
ax.plot(np.cos(theta),np.sin(theta),"blue",linewidth=2,label="x²+y²=1")

ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.set_aspect("equal")
ax.set_title(f"Monte Carlo pi Estimation (N= {N})\n pi = {np.round(np.pi,5)}",fontsize=13)
ax.legend(loc="lower left")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.tight_layout()
plt.show()

#convergence plot : pi estimate as N grows
cumulative_inside= np.cumsum(inside)
n_values = np.arange(1,N+1)
pi_running = 4*cumulative_inside/ n_values
fig2,ax2=plt.subplots(figsize=(9,5))
ax2.plot(n_values,pi_running,color="blue",linewidth=1,label="Estimated pi")
ax2.axhline(y=np.pi,color="red",linewidth=1.5,linestyle="--",label="True pi = 3.14159")
ax2.set_xlim(1,N)
ax2.set_ylim(2.5,4.0)
ax2.set_xlabel("Number of Points(N)")
ax2.set_ylabel("Estimated pi")
ax2.set_title("Convergence of Monte Carlo Pi estimate")
ax2.legend()
ax2.grid(True , alpha = 0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Number of random points
N = 10000

# Generate random points in the first octant
x = np.random.uniform(0, 1, N)
y = np.random.uniform(0, 1, N)
z = np.random.uniform(0, 1, N)

# Check whether the points lie inside the sphere
inside = x**2 + y**2 + z**2 <= 1

# Estimate pi
pi_estimate = 6 * np.sum(inside) / N

print(f"Points inside  : {np.sum(inside)}")
print(f"Points outside : {N - np.sum(inside)}")
print(f"Estimated pi   : {pi_estimate:.6f}")
print(f"True pi        : {np.pi:.6f}")
print(f"Error          : {pi_estimate - np.pi:.6f}")

# -----------------------------
# 3D Scatter Plot
# -----------------------------

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

# Transparent sphere surface
u = np.linspace(0, np.pi/2, 40)
v = np.linspace(0, np.pi/2, 40)

xs = np.outer(np.sin(u), np.cos(v))
ys = np.outer(np.sin(u), np.sin(v))
zs = np.outer(np.cos(u), np.ones_like(v))

ax.plot_surface(
    xs, ys, zs,
    color='cyan',
    alpha=0.15,
    edgecolor='gray',
    linewidth=0.3
)
ax.scatter(
    x[inside], y[inside], z[inside],
    color='green',
    s=4,
    label='Inside'
)

ax.scatter(
    x[~inside], y[~inside], z[~inside],
    color='red',
    s=4,
    label='Outside'
)

ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.set_zlim(0,1)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

ax.set_title(f"Monte Carlo π Estimation (Sphere)\nN = {N}")

ax.legend()

plt.tight_layout()
plt.show()

# -----------------------------
# Convergence Plot
# -----------------------------

cumulative_inside = np.cumsum(inside)

n_values = np.arange(1, N+1)

pi_running = 6 * cumulative_inside / n_values

fig2, ax2 = plt.subplots(figsize=(9,5))

ax2.plot(
    n_values,
    pi_running,
    color="blue",
    linewidth=1,
    label="Estimated π"
)

ax2.axhline(
    np.pi,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label="True π"
)

ax2.set_xlim(1, N)
ax2.set_ylim(2.5, 4)

ax2.set_xlabel("Number of Points (N)")
ax2.set_ylabel("Estimated π")
ax2.set_title("Convergence of Monte Carlo π Estimate (Sphere)")
ax2.grid(alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np 
import matplotlib.pyplot as plt 
 
num = 100000  # Using a higher number of samples for better accuracy 
prof = int(input("Please enter how much profit you want to earn (in Rs): ")) 
 
# Milk: 2 cups (1 cup = 0.25 liters) 
milk_per_cup = 0.25 
total_milk = 2 * milk_per_cup 
milk_cost = np.random.uniform(18, 20, num) 
 
# Eggs: 2 eggs 
egg_costs = np.random.choice([2, 2.5, 3], size=(num, 2), p=[0.2, 0.5, 0.3]) 
egg_costs = np.sum(egg_costs, axis=1) 
 
# Sugar: 100 grams (0.1 kg) 
sugar_kg = 0.1 
sugar_costs = np.random.uniform(20, 23, num) 
 
# Bread: 2 slices (Normal distribution for 12 slices) 
bread_slices = 2 
bread_costs = np.random.normal(25, 1, num) 
cost_per_slice = bread_costs / 12 
total_bread_cost = cost_per_slice * bread_slices 
 
# Almonds: 100 grams (0.1 kg) 
almonds_kg = 0.1 
almond_costs = np.random.choice([500, 700], size=num, p=[0.7, 0.3])

# Calculating total cost for each iteration 
total_costs = (total_milk * milk_cost) + egg_costs + (sugar_kg * 
sugar_costs) + total_bread_cost + (almonds_kg * almond_costs) 
 
# Mean of total costs to find the expected cost of pudding 
expected_cost = np.mean(total_costs) + prof 
print(f"Expected cost of pudding including profit using Monte Carlo simulation: Rs {expected_cost:.2f}") 
 
# Plotting the histogram 
plt.figure(figsize=(12, 6)) 
plt.hist(total_costs, bins=50, color='blue', edgecolor='black') 
plt.axvline(x=expected_cost - prof, color='red', linestyle='--', 
label=f'Expected Cost without Profit: Rs {expected_cost - prof:.2f}') 
plt.axvline(x=expected_cost, color='green', linestyle='--', 
label=f'Expected Cost with Profit: Rs {expected_cost:.2f}') 
plt.title('Distribution of Pudding Costs') 
plt.xlabel('Cost (Rs)') 
plt.ylabel('Frequency') 
plt.legend() 
plt.grid(True) 
plt.show()
```
